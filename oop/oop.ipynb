#### Exercise 1: Book Class
##### Create a Book class to represent books in a library.

```python
class Book:
    def __init__ (self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def info(self):
        print(f"{self.title} by {self.author}, {self.pages} pages")

    def is_long(self):
        return self.pages > 300
          
# create an instance of the Book class
book1 = Book("Harry Potter", "J.K. Rowling", 450)
book2 = Book("The Hobbit", "J.R.R. Tolkien", 250)

# call methods on the instances
book1.info()
book2.info()
print(book1.is_long())
print(book2.is_long()) 
```

#### Exercise 2: Student Class
##### Create a Student class to track student information and grades.

```python
class Student:
    def __init__ (self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def passed(self):
        return self.grade >= 50
        
    def letter_grade(self):
        if self.grade >= 80:
            return "A"
        elif self.grade >= 70:
            return "B"
        elif self.grade >= 60:
            return "C" 
        elif self.grade >= 50:
            return "D"
        else:
            return "F"
    
    def display(self):
        status = "passed" if self.passed() else "failed"
        print(f"\n----Student Info----")
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Grade: {self.grade}")
        print(f"Letter Grade: {self.letter_grade()}")
        print(f"Status: {status}")
    

# create an instance of the Student class
student1 = Student("Alice", 20, 85)
student2 = Student("Bob", 19, 45)

# call methods on the instances
print(student1.passed())
print(student2.passed())


student1.display()
student2.display()
```
