#### Exercise 1: Book Class 📚
##### Create a Book class to represent books in a library.

```python
class Book:
    def __init__ (self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def info(self):
        print(f"{self.title} by {self.author}, {self.pages} pages")

    def is_long(self):
        if self.pages > 300:
            return True
        else:
            return False

# create an instance of the Book class
book1 = Book("Harry Potter", "J.K. Rowling", 450)
book2 = Book("The Hobbit", "J.R.R. Tolkien", 250)

# call methods on the instances
book1.info()
book2.info()
print(book1.is_long())
print(book2.is_long()) 
```

#### Exercise 2: Student Class 🎓
##### Create a Student class to track student information and grades.

```python
class Student:
    def __init__ (self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def passed(self):
        if self.grade >= 50:
            return True
        else:
            return False
    
    def letter_grade(self):
        if self.grade >= 80:
            print("Grade: A")
        elif self.grade >= 70:
            print("Grade: B")
        elif self.grade >= 60:
            print("Grade: C")
        elif self.grade >= 50:
            print("Grade: D")
        else:
            print("Grade: F")
    
    def display(self):
        print(f"Name: {self.name}, Age: {self.age}, Grade: {self.grade}, Passed: {self.passed()}, Letter Grade: {self.letter_grade()}")

# create an instance of the Student class
student1 = Student("Alice", 20, 85)
student2 = Student("Bob", 19, 45)

# call methods on the instances
print(student1.passed())
print(student2.passed())

student1.letter_grade()
student2.letter_grade()

student1.display()
student2.display()
```
