#### Task 1: Product Inventory System 
##### Create a system to manage product inventory with stock tracking.

```python
# Task 1:
# Manage product information and stock levels using OOP principles.

class Product:
    def __init__ (self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def restock(self, quantity):
        self.stock += quantity
        print(f"Restocked {quantity} units of {self.name}. New stock: {self.stock}")

    def sell(self, quantity):
        result = self.stock >= quantity

        if result:
            self.stock = self.stock - quantity
            print(f"Sold {quantity} units of {self.name}. New stock: {self.stock}")
        else:
            print(f"Insufficient stock to sell {quantity} units of {self.name}. Current stock: {self.stock}")

    def get_value(self):
        return self.price * self.stock

    def is_low_stock(self):
        return self.stock < 10

    def display_info(self):
        print(f"\n--- Product Information ---")
        print(f"Name: {self.name}")
        print(f"Price: ${self.price:.2f}")
        print(f"Stock: {self.stock} units")
        print(f"Total Value: ${self.get_value():.2f}")
        if self.is_low_stock():
            print("Warning: Low stock!")
        else:
            print("Stock level is sufficient.")

# Example usage
laptop = Product("Laptop", 899.99, 15)
mouse = Product("Mouse", 25.50, 5)

# Display initial product info
laptop.display_info()
mouse.display_info()

# Sell some products
print("\n--- Sales ---")
laptop.sell(3)
mouse.sell(10)

# Test restocking
print("\n--- Restocking ---")
mouse.restock(20)

# Display updated info
print("\n--- Updated Inventory ---")
laptop.display_info()
mouse.display_info()
```

#### Task 2: Bank Account Manager
##### Create a banking system to manage customer accounts with transactions.

```python
# Task 2:
# Simulates basic banking operations using OOP.

class BankAccount:
    def __init__ (self, account_holder, account_number, initial_balance = 0, transactions = []):
        self.account_holder = account_holder
        self.account_number = account_number
        self.initial_balance = initial_balance
        self.transactions = transactions

    def deposit(self, amount):
        self.initial_balance += amount
        self.transactions.append(f"Deposited: ${amount:.2f}")
        print(f"Success! Deposited ${amount:.2f} into {self.account_holder} account. New balance: ${self.initial_balance:.2f}")

    def withdraw(self, amount):
        if amount <= self.initial_balance:
            self.initial_balance -= amount
            self.transactions.append(f"Withdrew: ${amount:.2f}")
            print(f"Success! Withdrew ${amount:.2f} from {self.account_holder} account. New balance: ${self.initial_balance:.2f}")
            return True
        else:
            print(f"Failed! Insufficient funds to withdraw ${amount:.2f} from {self.account_holder} account. Current balance: ${self.initial_balance:.2f}")
            return False
    
    def transfer(self, amount, recipient_account):
        if self.withdraw(amount):
            recipient_account.initial_balance += amount
            self.transactions.append(f"Transferred: ${amount:.2f} to {recipient_account.account_holder}")
            return True
        else:
            print(f"Failed! Insufficient funds to transfer ${amount:.2f}. Current balance: ${self.initial_balance:.2f}")
            return False
        
    def get_balance(self):
            return self.initial_balance

    def print_statement(self):
            print(f"\n---Account statement with transaction history---")
            print(f"Account Holder: {self.account_holder}")
            print(f"Account Number: {self.account_number}")
            print(f"Current Balance: ${self.initial_balance:.2f}")
            print("\n---Transaction History---")
            for transaction in self.transactions:
                print(transaction)

# Create accounts
alice_account = BankAccount("Alice Johnson", "ACC001", 1000)
bob_account = BankAccount("Bob Smith", "ACC002", 500)

# Perform transactions:
# Test deposits
print("--- Deposits ---")
alice_account.deposit(500)
bob_account.deposit(200)

# Test withdrawals
print("\n--- Withdrawals ---")
alice_account.withdraw(200)
bob_account.withdraw(1000)
  
# Test transfers
print("\n--- Transfers ---")
alice_account.transfer(300, bob_account)
 
# Print statements
print("\n--- Account Statements ---")
alice_account.print_statement()
bob_account.print_statement()
```
