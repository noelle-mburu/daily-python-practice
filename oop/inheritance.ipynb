#### Task: Book and EBook
##### Create a parent class Book and a child class EBook.

```python
# Inheritance with books.
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def info(self):
        print(f"'{self.title}' by {self.author}, {self.pages} pages")
    
    def is_long(self):
        return self.pages > 300

class EBook(Book):
        def __init__(self, title, author, pages, file_size):
            super().__init__(title, author, pages)
            self.file_size = file_size  # in MB
        
        def download_info(self):
            print(f"'{self.title}' is an eBook with a file size of {self.file_size} MB")

# Create instances
book1 = Book("The Hobbit", "J.R.R. Tolkien", 250)
ebook1 = EBook("Harry Potter", "J.K. Rowling", 450, 2.5)

# Test the classes
print("--- Regular Book ---")
book1.info()
print(f"Is long: {book1.is_long()}")

print("\n--- EBook ---")
ebook1.info()
print(f"Is long: {ebook1.is_long()}")
ebook1.download_info()
```

#### Task: BankAccount and SavingsAccount
##### Create a parent class BankAccount and a child class SavingsAccount.

```python
# Inheritance with bank accounts.
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited ${amount:.2f}. New balance: ${self.balance:.2f}")

    def get_balance(self):
        return self.balance

class SavingsAccount(BankAccount):
    def __init__(self, account_holder, balance, interest_rate):
        super().__init__(account_holder, balance)
        self.interest_rate = interest_rate  # as a percentage

    def add_interest(self):
        interest = self.balance * self.interest_rate
        print(f"Added interest: ${interest:.2f}")
        self.deposit(interest)

# Create accounts
regular_account = BankAccount("Alice", 1000)
savings_account = SavingsAccount("Bob", 5000, 0.05)  # 5% interest rate

# Perform transactions:
# Test regular account deposits
print("\n--- Regular Bank Account ---")
regular_account.deposit(500)

# Test balance retrieval
print(f"Current balance in regular account is ${regular_account.get_balance():.2f}")

# Test savings account deposits
print("\n--- Savings Account ---")
savings_account.deposit(1000)

# Test interest addition
print(f"Savings balance before interest: ${savings_account.get_balance():.2f}")
savings_account.add_interest()
print(f"Savings balance after interest: ${savings_account.get_balance():.2f}")
```
