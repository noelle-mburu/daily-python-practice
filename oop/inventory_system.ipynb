#### Task: Product Inventory System 
##### Create a system to manage product inventory with stock tracking.

```python
# Task:
# Manage product information and stock levels using OOP principles.

class Product:
    def __init__ (self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def restock(self, quantity):
        self.stock += quantity
        print(f"Restocked {quantity} units of {self.name}. New stock: {self.stock}")

    def sell(self, quantity):
        result = self.stock >= quantity

        if result:
            self.stock = self.stock - quantity
            print(f"Sold {quantity} units of {self.name}. New stock: {self.stock}")
        else:
            print(f"Insufficient stock to sell {quantity} units of {self.name}. Current stock: {self.stock}")

    def get_value(self):
        return self.price * self.stock

    def is_low_stock(self):
        return self.stock < 10

    def display_info(self):
        print(f"\n--- Product Information ---")
        print(f"Name: {self.name}")
        print(f"Price: ${self.price:.2f}")
        print(f"Stock: {self.stock} units")
        print(f"Total Value: ${self.get_value():.2f}")
        if self.is_low_stock():
            print("Warning: Low stock!")
        else:
            print("Stock level is sufficient.")

# Example usage
laptop = Product("Laptop", 899.99, 15)
mouse = Product("Mouse", 25.50, 5)

# Display initial product info
laptop.display_info()
mouse.display_info()

# Sell some products
print("\n--- Sales ---")
laptop.sell(3)
mouse.sell(10)

# Test restocking
print("\n--- Restocking ---")
mouse.restock(20)

# Display updated info
print("\n--- Updated Inventory ---")
laptop.display_info()
mouse.display_info()
```
