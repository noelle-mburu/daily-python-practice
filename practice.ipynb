```python
## Lists
## They are: Ordered, mutable, and allow duplicates
fruits = ["apple", "banana", "cherry"]
print(fruits[1])

# Using indexing to change an element
fruits[1] = "mango"
print(fruits)

# Adding elements to a list. Adds to the end of the list
fruits.append("melon")
print(fruits)

# Inserting an element at a specific position
fruits.insert(1, "coconut")
print(fruits)

# Looping through a list
for matunda in fruits:
    print(matunda)

# Copying a list
new_fruits = fruits.copy()
print(new_fruits)

# Slicing a list
print(fruits[1:3])
```

```python
# Dictionaries
# They are: ordered, mutable, and indexed by keys
person = {
    "name": "John",
    "age": 28,
    "city": "New York"
}

# Accessing values in a dictionary using keys NOT indexes
print(person["name"])

# Modifying a value in a dictionary
person["age"] = 37
print(person["age"])

# Adding a new key-value pair
person["email"] = "bobinsky@gmail.com"
print(person["email"])

# Looping through a dictionary
for key in person:
    print(key, ":", person[key])

for key in person:
    print(key)
```

```python
# Conditional Statements
temperature = 15

if temperature > 25:
    print("It's a hot day!")
else:
    print("It's not a hot day")


if temperature > 25:
    print("It's hot today!")
elif temperature > 10:
    print("It's a warm day!")
else:
    print("It's a cold day")

score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: D")

age = 22
has_ticket = False

if  age >= 18:
    if has_ticket:
        print("Entry allowed")
    else:
        print("Entry denied: No ticket")
else:
    print("Entry denied: Underage")

# Using and/or operators

if age >= 18 and has_ticket:
    print("Entry allowed")
else:
    print("Entry denied")
```

```python
# Using range() – Counting with for Loops

for i in range(4):
    print(i)

for i in range(5,8):
    print(i)

# Step by 2. The third value in range(start, stop, step) controls the steps.

for i in range(10,16,2):
    print(i)

# Skip with continue

for i in range(16,22,2):
    if i == 18:
        continue
    print(i)

# Exit loop with break

for i in range(4):
    if i == 3:
        break
    print(i)

# Loop With else

for i in range(3):
    print(i)
else:
    print("Loop completed")
```

```python
products = {
    "Laptop": 899.99,
    "Mouse": 25.50,
    "Keyboard": 159.99,
    "Monitor": 299.00
}

# Task 1: Print each product and its price

print("\n====All products and prices====")
for name, price in products.items():
    print(f"{name} : ${price:.2f}")

# Task 2: Find the total value of all products

print("\n==== Total value of all products ====")
total_value = sum(products.values())
print(f"Total values: ${total_value}")

# Task 3: Find the most expensive product

print("\n==== Most expensive product ====")
max_price = max(products.values())

for name, price in products.items():
    if price == max_price:
        most_expensive_name = name

print(f"Most expensive product: {most_expensive_name} at ${max_price:.2f}")
```

```python
# Functions with Multiple Parameters

def add(a,b):
    result = a + b
    print("Sum :", result)

add(7,3)

def total(c,d):
    price = c * d
    print("Total price :", price)

total(12,3)

def multiply(x, y):
    return x * y

result = multiply(3,4)
print(result)
```

```python
# Function to check if a number is even:
def is_even(n):
    return n % 2 == 0

print(is_even(4))   # True
print(is_even(9))   # False

# 0 == even number
# 1 == odd number
# % is modulus operator that gives the remainder of a division
# // is floor division that gives the integer/whole number part of a division
```
