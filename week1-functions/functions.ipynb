#### Task 1: Data Quality Report Generator 
##### Scenario: You're given a list of data values and need to create a report about data quality.
##### Goal: Create 3 functions that work together.
##### Functions:
- `count_missing_values()`: Counts None and empty string values in a dataset
- `calculate_completeness_percentage()`: Calculates percentage of non-missing values
- `generate_data_quality_report()`: Creates a complete data quality report with all metrics

```python
# Task 1.1: 
# Function that takes a list of values and returns the count of how many values are None or empty strings.

def count_missing_values(data_list):
    """Counts the missing values in a list"""
    return data_list.count(None) + data_list.count("")

total_missing_values = count_missing_values([25, 30, None, 45, "", 50, 35, None, 40, 28, 32, "", 38, 42, 29])
print("Total missing values:", total_missing_values)


# Task 1.2:
# Function to calculate what percentage of the data is NOT missing.

def calculate_completeness_percentage(data_list):
    """Calculates the percentage of non-missing values in a list"""
    missing_values = count_missing_values(data_list)
    return (len(data_list) - missing_values) / len(data_list) * 100

data_completeness_percentage = calculate_completeness_percentage([25, 30, None, 45, "", 50, 35, None, 40, 28, 32, "", 38, 42, 29])
print("data_completeness_percentage:", data_completeness_percentage)


# Task 1.3:
# Function to create a full report about the data quality.

customer_ages = [25, 30, None, 45, "", 50, 35, None, 40, 28, 32, "", 38, 42, 29]

def generate_data_quality_report(data_list, column_name):
    """Generates a data quality report for a given list"""
    total_records = len(data_list)
    missing_values = count_missing_values(data_list)
    complete_values = total_records - missing_values
    completeness_percentage = calculate_completeness_percentage(data_list)


    print("=== Data Quality Report ===")
    print(f"Column: {column_name}")
    print(f"Total records: {total_records}")
    print(f"Missing values: {missing_values}")
    print(f"Complete values: {complete_values}")
    print(f"Completeness percentage: {completeness_percentage:.2f}%")
    
    
data_quality_report = generate_data_quality_report(customer_ages, "Customer Ages")
```

#### Task 2: Data Transformation Pipeline 
##### Scenario: You received messy product data and need to clean it before analysis.
##### Goal: Create 4 functions that clean different aspects of the data.
##### Functions:
- `clean_product_name()`: Removes special characters and standardizes format
- `standardize_price()`: Converts price strings to float numbers
- `categorize_product_by_price()`: Categorizes products by price range  
- `process_product()`: Combines all functions to process complete product data

```python
# Task 2.1:
# Function that processes raw product data by cleaning names and standardizing prices.

def clean_product_name(product_name):
    """Cleans product names by removing special characters and extra spaces."""
    cleaned = product_name.strip()
    cleaned = cleaned.replace("#", "")
    cleaned = cleaned.replace("@", "")
    cleaned = cleaned.title()
    cleaned = " ".join(cleaned.split())
    return cleaned


raw_products = [
    ("  laptop  PRO###", "$899.99"),
    ("wireless MOUSE  ", "$25.50"),
    ("mechanical KEYBOARD@@", "$159.99"),
    ("USB  cable  ", "$12"),
    ("gaming HEADSET##", "$299")
]

""" Create a loop that separates raw names and prices, applies the cleaning function, and prints results """

for raw_name, raw_price in raw_products:
    cleaned_name = clean_product_name(raw_name)
    print("\n--- Product ---")
    print(f"Raw_name: {raw_name}")
    print(f"Raw price: {raw_price}")
    print(f"Cleaned name: {cleaned_name}")


# Task 2.2:
# Function that converts product prices to float

def standardize_price(price_string):
    """Converts price strings to float numbers."""
    cleaned = price_string.replace("$", "")
    return float(cleaned)


raw_products = [
    ("  laptop  PRO###", "$899.99"),
    ("wireless MOUSE  ", "$25.50"),
    ("mechanical KEYBOARD@@", "$159.99"),
    ("USB  cable  ", "$12"),
    ("gaming HEADSET##", "$299")
]

""" Loop through products, clean names, standardize prices, and print results """

for raw_name, raw_price in raw_products:
    cleaned_price = standardize_price(raw_price)
    print(f"\n--- Product price---")
    print(f"Raw price: {raw_price}")
    print(f"Standardized price: $ {cleaned_price:.2f}")


# Task 2.3:
# Function to categorize products based on price

def categorize_product_by_price(product_price):
    """Converts price strings to float numbers."""
    cleaned = product_price.replace("$", "")
    cleaned_price = float(cleaned)

    """if statement to categorize products based on price."""
    if cleaned_price < 50:
        category = "Budget"
    elif 50 <= cleaned_price < 200:
        category = "Mid-range"
    else:
        category = "Premium"
    return category

raw_products = [
    ("  laptop  PRO###", "$899.99"),
    ("wireless MOUSE  ", "$25.50"),
    ("mechanical KEYBOARD@@", "$159.99"),
    ("USB  cable  ", "$12"),
    ("gaming HEADSET##", "$299")
]

for raw_name, raw_price in raw_products:
    cleaned_price = float(raw_price.replace("$", ""))
    categories = categorize_product_by_price(raw_price)
    print (f"\n--- Product categories---")
    print(f"Cleaned price: $ {cleaned_price:.2f}")
    print(f"Category: {categories}")


# Task 2.4:
# Function to process raw product data by cleaning names, standardizing prices, categorizing products and returning a structured dictionary.

def process_product(raw_name, raw_price):
    """Processes raw product data and returns a structured dictionary."""
    cleaned_name = clean_product_name(raw_name)
    cleaned_price = standardize_price(raw_price)
    categories = categorize_product_by_price(raw_price)
    result = {
        "name": cleaned_name,
        "price": cleaned_price,
        "category": categories
    }
    return result

raw_products = [
    ("  laptop  PRO###", "$899.99"),    
    ("wireless MOUSE  ", "$25.50"),
    ("mechanical KEYBOARD@@", "$159.99"),
    ("USB  cable  ", "$12"),
    ("gaming HEADSET##", "$299")
]


for raw_name, raw_price in raw_products:
    processed_products = process_product(raw_name, raw_price)
    print("\n --- Processed Product ---")
    print(processed_products)
```
