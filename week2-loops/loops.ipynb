#### Task 1: Sales Report Generator
##### Scenario: You have daily sales data and need to generate a summary report.
##### Goal: Analyze daily sales data and generate insights.

```python
# Task 1.1: Calculate total sales

daily_sales = [150.50, 200.00, 175.25, 300.00, 125.75, 180.00, 220.50, 195.00, 240.75, 210.00]

Total_sales = sum(daily_sales)
print(f"Total Sales: ${Total_sales}")

# Task 1.2:
# Calculate average daily sales

average_sales = Total_sales / len(daily_sales)
print(f"\nAverage Daily Sales: ${average_sales:.2f}")

# Task 1.3:
# Find best and worst sales days

for day, sales in enumerate(daily_sales, start=1):
    if sales == max(daily_sales):
        print(f"\nBest Sales Day is Day {day} with ${sales}")
    elif sales == min(daily_sales):
        print(f"\nWorst Sales Day is Day {day} with ${sales}")

# Task 1.4:
# Count number of days with sales above $200

count = 0
target = 200

for sales in daily_sales:
    if sales > target:
        count += 1
        
print(f"\nDays with sales above ${target}: {count}")
```

#### Task 2: Customer Data Processor
##### Scenario: You have customer records that need cleaning and analysis.
##### Goals: Clean and analyze customer data from multiple sources.

```python
# Task 2.1: Clean customer data

customers = [
    {"name": "  alice JOHNSON  ", "email": "ALICE@EMAIL.COM", "purchases": 5, "total_spent": 450.00},
    {"name": "bob SMITH", "email": "bob@email.com  ", "purchases": 12, "total_spent": 890.50},
    {"name": "CHARLIE Brown  ", "email": "  charlie@email.com", "purchases": 3, "total_spent": 125.00},
    {"name": "diana WILSON", "email": "DIANA@EMAIL.COM", "purchases": 8, "total_spent": 670.25},
    {"name": "  EVE Davis  ", "email": "eve@email.com  ", "purchases": 15, "total_spent": 1200.00},
]

clean_customers = []

for customer in customers:
    name = customer["name"].strip().title()
    email = customer["email"].strip().lower()
    purchases = customer["purchases"]
    total_spent = customer["total_spent"]
    
    clean_customer = {
        "name": name,
        "email": email,
        "purchases": purchases,
        "total_spent": total_spent
    }

    clean_customers.append(clean_customer)


print(f"Cleaned {len(clean_customers)} customer records")

for customer in clean_customers:
    print(customer)


# Task 2.2: Calculate average purchase value

for customer in clean_customers:
    avg_price = customer["total_spent"] / customer["purchases"]
    print(f"\n{customer['name']}'s average purchase value: ${avg_price:.2f}")


# Task 2.3: Categorize customers

VIP_count = 0
Regular_count = 0
New_count = 0

for customer in clean_customers:
    if customer["total_spent"] >= 1000:
        category = "VIP"
        VIP_count += 1
    elif customer["total_spent"] >= 500:
        category = "Regular"
        Regular_count += 1
    else:
        category = "New"
        New_count += 1

    customer["category"] = category


print("\nCustomer Categories Summary:")
print(f"{VIP_count} VIP customers \n{Regular_count} Regular customers \n{New_count} New customers")

print("\n=== Customers with Categories ===")
for customer in clean_customers:
    print(f"{customer['name']}: {customer['category']}")

# Task 2.4: Find top spender
# Loop through customers and find who spent the most
# Print: "Top spender: [name] with $[amount]"

for customer in clean_customers:
    top_spender = max(clean_customers["total_spent"])

    if customer["total_spent"] == top_spender:
        print(f"\nTop spender: {customer['name']} with ${top_spender}")




# Task 2.5: Generate email list
# Create a list of just the cleaned email addresses
# Use a for loop (not list comprehension yet!)
# Print: "Generated email list with [X] addresses"
# Print the list
```
